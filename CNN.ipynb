```python
import keras 
import numpy as np
from keras.utils import to_categorical
from keras.layers import Conv2D,Dense,MaxPooling2D,Flatten
from keras import models
import matplotlib.pyplot as plt
```

```python
(x_train,y_train),(x_test,y_test)=keras.datasets.mnist.load_data()
```

```python
y_test.shape
```

```python
x_train.reshape((x_train.shape[0],x_train.shape[1],x_train.shape[2],1))
x_test.reshape((x_test.shape[0],x_test.shape[1],x_test.shape[2],1))
```

```python
x_train=x_train[:10000].astype('float32')/255
x_test=x_test[:1000].astype('float32')/255
```

```python
y_train=to_categorical(y_train[:10000])
y_test=to_categorical(y_test[:1000])
```

```python
model=models.Sequential([
    Conv2D(32,kernel_size=(3,3),activation="relu",input_shape=(28,28,1)),
    MaxPooling2D(2,2),
    Conv2D(64,kernel_size=(3,3),activation="relu"),
    MaxPooling2D(2,2),
    Flatten(),#Crucial when transitioning from CNN layers to Dense layer
    Dense(64,activation="relu"),
    Dense(10,activation="softmax")
])

model.summary()
```

```python
model.compile(optimizer="adam",loss="categorical_crossentropy",metrics=["accuracy"])
```

```python
from keras import callbacks
earlystopping=callbacks.EarlyStopping(monitor="val_loss",mode="min",patience=5,restore_best_weights=True)
```

```python
history=model.fit(x_train,y_train,batch_size=128,epochs=25,callbacks=[earlystopping])
```
